## Q05

キーボードからいくつかの整数（点数）を半角スペース区切りで入力する。入力された整数の**平均値**が60以上なら `合格`、40以上60未満なら `再テスト`、40未満なら `不合格` と出力するプログラムを作成せよ。

例えば、キーボードから
```
50 60 70 80
```
が入力されたら、平均値が65であるため `合格` を出力し、
```
40 45 50
```
が入力されたら、平均値が45であるため `再テスト` を出力する。

プログラムの先頭に
```
mylist = list(map(int, input().split()))
```
と書くと、キーボードから半角スペース区切りで入力された整数が、リスト型変数 `mylist` に代入される。

キーボードからは1つ以上の整数が入力されることを前提としてよい。

**ヒント：** 平均値は「合計 ÷ 個数」で求められます。リストの要素の合計を求める関数と、要素の個数を求める関数は？

**サンプルケース1**
- 入力値: `60 70 80`
- 出力値: `合格`

**サンプルケース2**
- 入力値: `30 50 45`
- 出力値: `再テスト`

**サンプルケース3**
- 入力値: `10 20 30 40`
- 出力値: `不合格`

````python
mylist = list(map(int, input().split()))
ave = sum(mylist) / len(mylist)
if ave >= 60:
    print("合格")
elif ave >= 40:
    print("再テスト")
else:
    print("不合格")
````

## Q08

キーボードから2つの整数を入力する。最初に入力した数を $a$、つぎに入力した数を $b$ とおく。このとき、$a$ が $b$ で割り切れるかどうかを調べるプログラムを作成せよ。割り切れるときは
```
aはbで割り切れます
```
割り切れないときは
```
aはbで割り切れません
```
と出力するようにせよ。ただし、$a$ と $b$ のどちらか一方でも0以下の場合は
```
無効な入力です
```
と出力するようにせよ。

例えば、キーボードから
```
4
2
```
が入力されたときは `4は2で割り切れます` が出力され、
```
3
5
```
が入力されたときは `3は5で割り切れません` が出力される。

**ヒント：** $b$ が0のときに割り算をするとエラーになります。0以下かどうかの判定を先に行いましょう。

**サンプルケース1**
- 入力値: 
```
10
5
```
- 出力値: `10は5で割り切れます`

**サンプルケース2**
- 入力値: 
```
5
9
```
- 出力値: `5は9で割り切れません`

**サンプルケース3**
- 入力値: 
```
4
4
```
- 出力値: `4は4で割り切れます`

**サンプルケース4**
- 入力値: 
```
8
0
```

- 入力値: 
```
-6
3
```
- 出力値: `無効な入力です`

````python
a = int(input())
b = int(input())
if a <= 0 or b <= 0:
    print("無効な入力です")
elif a % b == 0:
    print(f"{a}は{b}で割り切れます")
else:
    print(f"{a}は{b}で割り切れません")
````

## Q09

キーボードから正の整数 $a$ が入力されたとき、$1$ から $a$ までの整数のうち**平方数**（ある整数を2乗した数：1, 4, 9, 16, …）をリスト型変数として出力するプログラムを作成せよ。

例えば、$a=19$ が入力されたとき
```
[1, 4, 9, 16]
```
が出力される。
```
a = int(input())
```
のように書き始めると良い。

**ヒント：** $i$ を1から順に増やしていき、$i^2$ が $a$ 以下である間だけリストに追加していく方法が考えられます。

**サンプルケース1**
- 入力値: `19`
- 出力値: `[1, 4, 9, 16]`

**サンプルケース2**
- 入力値: `1`
- 出力値: `[1]`

**サンプルケース3**
- 入力値: `50`
- 出力値: `[1, 4, 9, 16, 25, 36, 49]`

````python
a = int(input())
mylist = []
i = 1
while i * i <= a:
    mylist.append(i * i)
    i += 1
print(mylist)
````

## Q10

list型変数
```
scores = [5, 1, 3, 9, 0, 1, 2, 8, 1, 9]
```
がある。キーボードから整数が入力されたとき、その整数が `scores` の中に何回出現するかを調べるプログラムを作成せよ。
例えば、キーボードから `9` が入力されたとき
```
9はscoresの中に2回出現します
```
と出力し、`scores` の中に一度も出現しない整数が入力されたときは、例えば `10` なら
```
10はscoresの中にありません
```
と出力するようにせよ。

プログラムはlist型変数 `scores` の要素や要素数が変更されても対応できるものにせよ。
```
a = int(input())
```
のように書き始めると良い。

**サンプルケース1**
- 入力値: `1`
- 出力値: `1はscoresの中に3回出現します`

**サンプルケース2**
- 入力値: `0`
- 出力値: `0はscoresの中に1回出現します`

**サンプルケース3**
- 入力値: `10`
- 出力値: `10はscoresの中にありません`

````python
a = int(input())
scores = [5, 1, 3, 9, 0, 1, 2, 8, 1, 9]
count = scores.count(a)
if count > 0:
    print(f"{a}はscoresの中に{count}回出現します")
else:
    print(f"{a}はscoresの中にありません")
````
